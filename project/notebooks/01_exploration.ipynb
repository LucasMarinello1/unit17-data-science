```python
import pandas as pd

sheets = pd.read_excel("../data/raw/online_retail_II.xlsx", sheet_name=None)
df = pd.concat(sheets.values(), ignore_index=True)

print("Hojas:", list(sheets.keys()))
print("Shape:", df.shape)
print(df.dtypes, "\n")
print("Periodo:", df["InvoiceDate"].min(), "->", df["InvoiceDate"].max())

# Calidad
print("\nNulos por columna:\n", df.isna().sum())
print("% filas sin Customer ID:", round(df["Customer ID"].isna().mean() * 100, 1))
print("Filas duplicadas exactas:", df.duplicated().sum())
cancel = df["Invoice"].astype(str).str.startswith("C")
print("Líneas de cancelación:", cancel.sum())
print("Cantidad <= 0:", (df["Quantity"] <= 0).sum())
print("Precio <= 0:", (df["Price"] <= 0).sum())

# Dimensiones
print("\nClientes únicos:", df["Customer ID"].nunique())
print("Productos únicos:", df["StockCode"].nunique())
print("Países:", df["Country"].nunique())
print(df["Country"].value_counts().head(5))

# Guardar en parquet para no releer el Excel
for col in ["Invoice", "StockCode", "Description", "Country"]:
    df[col] = df[col].astype("string")
df.to_parquet("../data/online_retail_ii.parquet", index=False)
print("Guardado OK")
```
